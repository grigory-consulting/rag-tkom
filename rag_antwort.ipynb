# Antwort mit RAG

Die Frage wird zuerst im P-12-Korpus gesucht. Nur die gefundenen Dokumentabschnitte werden dem LLM als Kontext übergeben.

```python
from common import llm
from common.corpus import load_corpus
from common import retrieval as R
```

```python
# Vorhandenes Hybrid-Retrieval des Projekts aufbauen
dokumente = load_corpus()
chunks = R.chunk_corpus(dokumente)
retriever = R.HybridRetriever(chunks)
```

```python
frage = "Wie hoch ist der Dauerbetriebsdruck der Hydraulikpumpe P-12?"

treffer = retriever.search(frage, k=3)
kontext = R.format_context(treffer)

for chunk, score in treffer:
    print(f"[{chunk.doc_id}] {chunk.title} ({score:.4f})")
```

```python
antwort = llm.complete(
    f"Frage:\n{frage}\n\nKontext:\n{kontext}",
    system=(
        "Sie sind ein Hydraulik-Experte. Antworten Sie ausschließlich "
        "anhand des Kontexts, knapp und konkret. Zitieren Sie die "
        "verwendete Quelle in eckigen Klammern."
    ),
    temperature=0.0,
    max_tokens=120,
)

print(antwort)
```
